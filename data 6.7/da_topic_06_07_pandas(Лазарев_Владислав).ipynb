## Контекст проекта
Ваш руководитель попросил подготовить аналитический отчёт по всем клиентам и покупкам магазина.
Данные лежат в трёх CSV-файлах. Используя Pandas, вы должны загрузить, объединить и проанализировать их.

**Цель урока:** Освоить загрузку данных, фильтрацию, группировку, объединение таблиц и агрегации в Pandas.

```python
import pandas as pd

customers = pd.read_csv("customers.csv")
transactions = pd.read_csv("transactions.csv")
products = pd.read_csv("products.csv")
```

---
### Задание 1: Первичный осмотр
Выведите первые 10 строк таблицы `customers`. Затем выведите типы данных всех колонок с помощью `.dtypes`.

```python
print(customers.head(10))
print(customers.dtypes)
```

---
### Задание 2: Описательная статистика
Используйте `.describe()` для таблицы `transactions`. Какое среднее количество товаров в одной покупке?

```python
print(transactions.describe())

print("Среднее количество товаров в покупке:",
      transactions["quantity"].mean())
```

---
### Задание 3: Фильтрация — Клиенты из Москвы
Отфильтруйте таблицу `customers` и оставьте только клиентов из Москвы (`location == 'Moscow'`).
Сколько их?

```python
moscow_customers = customers[customers["location"] == "Moscow"]

print(moscow_customers)
print("Количество клиентов из Москвы:", len(moscow_customers))
```

---
### Задание 4: Объединение таблиц (Merge)
Объедините `transactions` с `customers` по ключу `customer_id`, затем добавьте `products` по `product_id`.
Сохраните в `df`. Выведите `.shape` и `.head()`.

```python
df = transactions.merge(customers, on="customer_id")
df = df.merge(products, on="product_id")

print("Размер таблицы:", df.shape)
print(df.head())
```

---
### Задание 5: Создание нового столбца
Добавьте в `df` колонку `total_sum` = `price` * `quantity`.
Выведите 5 самых дорогих покупок (отсортируйте по `total_sum` по убыванию).

```python
df["total_sum"] = df["price"] * df["quantity"]

top_expensive = df.sort_values(
    by="total_sum",
    ascending=False
).head(5)

print(top_expensive)
```

---
### Задание 6: Группировка — Выручка по категориям
Сгруппируйте `df` по `category` и посчитайте суммарную выручку (`total_sum.sum()`) по каждой категории.
Отсортируйте по убыванию.

```python
revenue_by_category = df.groupby("category")["total_sum"].sum()

revenue_by_category = revenue_by_category.sort_values(
    ascending=False
)

print(revenue_by_category)
```

---
### Задание 7: Группировка — Средний чек по городам
Посчитайте средний чек (`total_sum.mean()`) для каждого города (`location`). В каком городе самый высокий средний чек?

```python
average_check = df.groupby("location")["total_sum"].mean()

print(average_check.sort_values(ascending=False))

print("Город с самым высоким средним чеком:",
      average_check.idxmax())
```

---
### Задание 8: Множественная агрегация
Используя `.agg()`, посчитайте для каждой категории товаров:
- Количество транзакций (`count`)
- Среднюю сумму (`mean`)
- Максимальную сумму (`max`)

```python
result = df.groupby("category")["total_sum"].agg(
    ["count", "mean", "max"]
)

print(result)
```

---
### Задание 9: ТОП-10 покупателей
Найдите ТОП-10 клиентов по общей сумме покупок. Выведите их `customer_id`, `location` и суммарную `total_sum`.

```python
top_10 = df.groupby(
    ["customer_id", "location"],
    as_index=False
)["total_sum"].sum()

top_10 = top_10.sort_values(
    by="total_sum",
    ascending=False
).head(10)

print("ТОП-10 покупателей:")
print(top_10)
```

---
### Задание 10: Анализ по времени
Преобразуйте колонку `date` в формат `datetime`. Добавьте колонку `month` (месяц покупки).
Посчитайте общую выручку по месяцам.

```python
# Ваш код здесь
df["date"] = pd.to_datetime(df["date"])

df["month"] = df["date"].dt.to_period("M")

monthly_revenue = df.groupby("month")["total_sum"].sum()

print("Общая выручка по месяцам:")
print(monthly_revenue)
```

---
### Задание 11: Уникальные значения и Value Counts
1. Сколько уникальных городов в датасете?
2. Выведите распределение клиентов по полу (`gender`) с помощью `.value_counts()`.

```python
print("Количество уникальных городов:",
      customers["location"].nunique())

print("Распределение клиентов по полу:")
print(customers["gender"].value_counts())
```

---
### Задание 12 (Бонус): Кросс-таблица
Создайте кросс-таблицу (`pd.crosstab`), показывающую количество покупок для каждой комбинации `gender` и `category`.

```python
cross_table = pd.crosstab(
    df["gender"],
    df["category"]
)

print(cross_table)
```
